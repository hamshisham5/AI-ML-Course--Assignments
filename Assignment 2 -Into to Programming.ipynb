1. Write a function that takes a list of strings and a number K. The function should return a new list containing only the strings that have a length greater than K.
   - **Input:** `list=["apple", "pie", "banana"], K=3`
   - **Output:** `['apple', 'banana']`
   

```python
def filter_by_length(lst, k):
    # Solution 1: Using Filter
    result = list(filter(lambda x: len(x) > k, lst))

    # Solution 2: Using a For loop
    result = []
    for item in lst:
        if len(item) > k:
            result.append(item)

    # Solution 3: Using List Comprehension
    result = [item for item in lst if len(item) > k]
    
    return result

lst = ["apple", "pie", "banana"]
k = 3

filter_by_length(lst, k)
```

2. Write a function named `area()` that calculates the area of either a "circle" or a "rectangle". The function should take the shape name as the first argument, followed by the necessary dimensions (radius for circle; length and width for rectangle).
   - **Example Input:** `area("circle", 5)` OR `area("rectangle", 4, 6)`
   - **Example Output:** `Circle Area: 78.5`, `Rectangle Area: 24`

```python
# Solution 1: Using Default Parameters
def area(shape, dim1, dim2=None):
    if shape == "circle":
        pi = 22 / 7
        radius = dim1
        return 2 * pi * radius

    if shape == "rectangle":
        return dim1 * dim2

# Solution 2: Using *args (Multiple Arguments)
def area(shape, *dims):
    if shape == 'circle':
        pi = 22 / 7
        radius = dims[0]
        return 2 * pi * radius
    if shape == 'rectangle':
        return dims[0] * dims[1]

print(area('rectangle', 5, 2))
print(area('circle', 3))
```

4. Write a program to find the sum of all numbers in a list. To make it mid-level, ensure the program ignores any elements in the list that are not numbers (like strings or booleans).
   - **Example Input:** `[10, "apple", 20, True, 5]`
   - **Example Output:** `Total Sum: 35`

```python
def sum_nums(lst):
    # Solution 1: Using Filter
    only_num_list = list(filter(lambda x: type(x) in [int, float], lst))
    
    # Solution 2: Using a for loop
    only_num_list = []
    for item in lst:
        if type(item) in [int, float] :
            only_num_list.append(item)
    
    # Solution 3: Using list comprehension
    only_num_list = [item for item in lst if type(item) in [int, float]]
    
    int_sum = sum(only_num_list)
    return int_sum

sum_nums([10, "apple", 20, True, 5])
```

5.	Write a Python class Restaurant with attributes like menu_items, book_table, and customer_orders, and methods like add_item_to_menu, book_tables, and customer_order. (Note: Use dictionaries and lists to store the data.)
Perform the following tasks now:
-	Now add items to the menu.
-	Make table reservations.
-	Take customer orders.
-	Print the menu.
-	Print table reservations.
-	Print customer orders.


```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}  # Dictionary to store menu items and their prices
        self.book_table = []  # List to store table reservations
        self.customer_orders = {}  # Dictionary to store customer orders

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, customer_name, table_number):
        self.book_table.append({"customer_name": customer_name, "table_number": table_number})

    def customer_order(self, customer_name, order_items):
        if customer_name not in self.customer_orders:
            self.customer_orders[customer_name] = []
        self.customer_orders[customer_name].extend(order_items)

    def print_menu(self):
        print("Menu:")
        for item, price in self.menu_items.items():
            print(f"{item}: ${price}")

    def print_table_reservations(self):
        print("Table Reservations:")
        for reservation in self.book_table:
            print(f"Customer: {reservation['customer_name']}, Table: {reservation['table_number']}")

    def print_customer_orders(self):
        print("Customer Orders:")
        for customer, orders in self.customer_orders.items():
            print(f"Customer: {customer}, Orders: {', '.join(orders)}")


# Example usage
restaurant = Restaurant()

# Add items to the menu
restaurant.add_item_to_menu("Pizza", 10)
restaurant.add_item_to_menu("Burger", 8)
restaurant.add_item_to_menu("Pasta", 12)

# Make table reservations
restaurant.book_tables("Alice", 1)
restaurant.book_tables("Bob", 2)

# Take customer orders
restaurant.customer_order("Alice", ["Pizza", "Pasta"])
restaurant.customer_order("Bob", ["Burger"])

# Print the menu
restaurant.print_menu()

# Print table reservations
restaurant.print_table_reservations()

# Print customer orders
restaurant.print_customer_orders()
```

class Bike that is composed of the following attributes: 
-	name 
-	price 

Here are some of the methods that you need to add to your class Bike; 
-	def `__init__(self)`. 
-	Getter / setter for all attributes.


```python
class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name}, {self.price}"

    def __repr__(self):
        return f"Bike({self.name}, {self.price})"

    def get_name(self):
        return self.name

    def set_name(self, value):
        self.name = value

    def get_price(self):
        return self.price

    def set_price(self, value):
        self.price = value
```

class MainBikeRental that is composed of the following attributes: 
-	List of bikes 
-	List of Rentals 

Here are some of the methods that you need to add to your class MainBikeRental; 
-	def `__init__(self)`. 
-	Getter / setter for all attributes. 
-	**requestBike**: Takes a request from the customer for the number of bikes and check if the requested number is available or not if yes, function will deduct number_of_bikes from shop stock else it will print wait 10 mins. And you have to make sure that the age of the customer is over than 6 years. 
-	**returnBike**: Allows customers to return their bikes to the rental shop by taking rentalTime, num_of_bikes as parameters. And calculate the total bill cost by multiplying rentalTime * 40. Also, add the returned bikes to the stock store again 
-	**addBike**: adds bikes into the list 
-	**totalCost**: returns the total cost of bikes in the store 


```python
class MainBikeRental:
    def __init__(self):
        # Store all available bikes in the shop
        self.bikes = []
        # Track which customer has rented which bikes
        self.rentals = {}

    def requestBike(self, customer, num_bikes):
        """Allow a customer to rent bikes from the shop"""
        
        # Check if we have enough bikes available
        if num_bikes > len(self.bikes):
            return "Wait 10 Mins"
        
        # Check if customer is old enough to rent
        if customer.age < 6:
            return "Must be over 6 years old"
        
        # Get the bikes the customer wants to rent
        rented_bikes = self.bikes[:num_bikes]
        
        # Remove those bikes from available bikes
        self.bikes = self.bikes[num_bikes:]
        
        # Save the rental record
        self.rentals[customer.name] = rented_bikes
        
        return f"{num_bikes} Bikes rented to {customer.name}."

    def returnBike(self, customer, num_bikes, rental_time):
        """Allow a customer to return bikes and calculate the cost"""
        
        # Get the bikes this customer rented
        rented_bikes = self.rentals[customer.name]
        
        # Add the returned bikes back to the shop
        for _ in range(num_bikes):
            returned_bike = rented_bikes.pop()
            self.bikes.append(returned_bike)
        
        # Calculate total rental cost (40 per hour per bike)
        total_cost = rental_time * 40 * num_bikes
        
        return total_cost

    def addBike(self, bike):
        """Add a new bike to the shop"""
        self.bikes.append(bike)

    def totalCost(self):
        """Calculate the total value of all bikes in the shop"""
        total = 0
        for bike in self.bikes:
            total += bike.price
        return total


# Create the rental system
system = MainBikeRental()

# Add 10 bikes to the shop with increasing prices
for bike_number in range(10):
    bike_name = f"B{bike_number + 1}"
    bike_price = 500 + bike_number * 50
    system.addBike(Bike(bike_name, bike_price))

# Display initial shop status
print("Bikes:", system.bikes)
print("Rentals:", system.rentals)
print("Total Cost:", system.totalCost())

print("\n###########\n")

# Create a customer named Sameh, age 10
customer_sameh = Customer("Sameh", 10)

# Request 4 bikes
print(f"Number of bikes in the store: {len(system.bikes)}")
print(system.requestBike(customer_sameh, 4))
print(f"Remaining bikes in store: {len(system.bikes)}")
print(f"Current rentals: {system.rentals}\n\n")

# Return 2 bikes after 20 hours
print(f"Return cost: {system.returnBike(customer_sameh, 2, 20)}")
print(f"Remaining bikes in store after return: {len(system.bikes)}")
print(f"Updated rentals: {system.rentals}\n\n")

# Return 2 more bikes after 20 hours
print(f"Return cost: {system.returnBike(customer_sameh, 2, 20)}")
print(f"Remaining bikes in store after return: {len(system.bikes)}")
print(f"Updated rentals: {system.rentals}\n\n")
```

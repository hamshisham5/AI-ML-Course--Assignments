## Question 1: Normalize a 3x3 Array

```python
array1 = np.random.randint(1, 101, size=(3, 3))

smallest = np.min(array1)
largest = np.max(array1)

if largest == smallest:
    normalized = np.zeros((3, 3))
else:
    normalized = (array1 - smallest) / (largest - smallest)

print("Original array:")
print(array1)

print("Normalized array:")
print(normalized)
```

## Question 2: Find the Closest Value to X

```python
array2 = np.random.randint(1, 101, size=10)

print("Array:", array2)

x = float(input("Enter a number for X: "))

distances = np.abs(array2 - x)

closest_index = np.argmin(distances)
closest_value = array2[closest_index]

print("X =", x)
print("Closest value to X =", closest_value)
```

## Question 3: Second Largest Value in Each Column

```python
array3 = np.random.randint(1, 101, size=(5, 5))

sorted_columns = np.sort(array3, axis=0)

second_largest = sorted_columns[-2, :]

print("Original array:")
print(array3)

print("Second largest values:")
print(second_largest)
```

## Question 4: Rows with Even Sums

```python
array4 = np.random.randint(1, 21, size=(6, 4))

row_sums = np.sum(array4, axis=1)

even_row_indices = np.where(row_sums % 2 == 0)[0]

print("Original matrix:")
print(array4)

print("Row sums:")
print(row_sums)

print("Even row indices:")
print(even_row_indices)
```

## Question 5: Replace Multiples of 3 or 5 with -1

```python
array5 = np.random.randint(1, 51, size=30)

original_array5 = array5.copy()

multiples = (array5 % 3 == 0) | (array5 % 5 == 0)

array5[multiples] = -1

print("Original array:")
print(original_array5)

print("Modified array:")
print(array5)
```

## Question 6: Subtract Row Mean from Each Element

```python
array6 = np.random.randint(1, 31, size=(3, 3))

row_means = np.mean(array6, axis=1, keepdims=True)

result6 = array6 - row_means

print("Original array:")
print(array6)

print("Row means:")
print(row_means)

print("Final array:")
print(np.round(result6, 2))
```
